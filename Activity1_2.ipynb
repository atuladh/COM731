Read this csv into a dataframe called df_data 

```python
#Write your code here
df_data = pd.read_csv("Purchase_Data.csv")
```

```python
#Write your code here
df_data.dtypes
```

Slice the <b>first two columns</b> using .loc and store the result in a variable called result_1.

```python
#Write you code here
result_1 = df_data.loc[:, ["invoice_number", "invoice_time"]]
result_1
```

Slice the <b>first two rows</b> using .loc and store the result in a variable called result_2.

```python
#Write you code here
result_2 = df_data.loc[[1,2], :]
result_2
```

Slice the first four rows and the first five columns and store the result in a variable called result_3.

```python
#Write your code here
results_3 = df_data.iloc[0:4, 0:5]
results_3
```

```python
#Write your code here
result_4 = df_data.loc[[0,4,6], ["invoice_time", "price"]]
result_4
```
