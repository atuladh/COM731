```python
#import csv
import csv

#Create a variable named file_name and assign it the following: "user_entries.csv"
file_name = "user_entries.csv"

#Open the file for writing using with and  Python's open() function. The program should use the variable file_name. 
with open(file_name, mode = "w", encoding = 'UTF-8') as csv_file:
    #create writer object
    writer = csv.writer(csv_file)

    #write the header
    writer.writerow(["year", "artist", "song_title"])
    
    #loop to get user input
    for i in range(3):
        year = input("Enter year: ")
        artist = input("Enter artist: ")
        song_title = input("Enter song title: ")
        #write row to csv file
        writer.writerow([year, artist, song_title])
```
