```python
#To add prices of given number of samples
number = int(input("Enter total number of samples: "))
i = 0
total_price = 0
while i < number:
    price = int(input("Enter price of an item: "))
    total_price = total_price + price
    i = i + 1

print(f"Total price of {number} samples is GBP {total_price}")
```

```python
#Attendance counter
number = int(input("Enter total number of students: "))
i = 0
a = 0
roll = 1
ab = 0
while i < number:
    yes_or_no = (input(f"Enter attendance of Roll number {roll} (y/n): "))
    if yes_or_no == "y":
        a = a + 1
    roll = roll + 1
    i = i + 1
print(f"Total number of students present is {a}")
```

```python
#Attendance counter
number = int(input("Enter total number of students: "))
i = 0
a = 0
roll = 1
ab = 0
while i < number:
    yes_or_no = (input(f"Enter attendance of Roll number {roll} (y/n): "))
    if yes_or_no == "y":
        a = a + 1
        roll = roll + 1
        i = i + 1
    elif yes_or_no == "n":
        ab = ab + 1
        roll = roll + 1
        i = i + 1
    else:
        print("Enter valid letter only, y or n")
roll = roll + 1
i = i + 1
print(f"Total number of students present is {a}")
print(f"Total number of students absent is {ab}")
```

```python
#Attendance counter
num = int(input("Enter total number of students enrolled in class: "))
present_count = 0
for student_num in range(1, num + 1):
    response = input(f"Enter attendance of Roll number {student_num} (y/n): ")
    if response == "y":
        present_count = present_count + 1
percent = (present_count / num) * 100
print(f"Total number of students present is {present_count} out of {num} and in terms of percentage is {percent:.2f} %")
```
