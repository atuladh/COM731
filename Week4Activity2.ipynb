```python
#To add prices of given number of samples
number = int(input("Enter total number of samples: "))
i = 0
total_price = 0
while i < number:
    price = int(input("Enter price of an item: "))
    total_price = total_price + price
    i = i + 1

print(f"Total price of {number} samples is GBP {total_price}")
```
