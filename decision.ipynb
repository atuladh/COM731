Program to check if given angles of a traingle forms a legitimate traingle or not

```python
#Input the angles
angle1 = int(input("Enter the first angle of the triangle: "))
angle2 = int(input("Enter the second angle of the triangle: "))
angle3 = int(input("Enter the third angle of the triangle: "))
#Calculate the sum of the angles
angle_sum = angle1 + angle2 + angle3
#Determine if the angles form a triangle
if angle_sum == 180:
    print("The angles form a triangle.")
else:
    print("The angles do not form a triangle.")
```

```python
#Ask user to input the angles
angle1 = int(input("Enter the first angle of the triangle: "))
angle2 = int(input("Enter the second angle of the triangle: "))
angle3 = int(input("Enter the third angle of the triangle: "))

#Calculate the sum of the an gles
sum_of_angles = angle1 + angle2 + angle3

#Determine the type of triangle
if angle1 == angle2 == angle3:
    triangle = "equilateral"
elif angle1 == angle2 or angle2 == angle3 or angle1 == angle3:
    triangle = "isosceles"
elif (angle1 != angle2) and (angle2 != angle3) and (angle1 != angle3):
    triangle = "scalene"

#Display the type of triangle
if sum_of_angles == 180:
    print(f"The angles form a {triangle} triangle.")
else: 
    print("The angles do not form a triangle.")
```

BMI calculator

```python
#Ask for weight (kg) and height (cm)
#ask for unit of height and then value for height
unit1 = input("Which unit would you like to enter your height in? (cm or inches): ")
if unit1 == "cm":
    height = float(input("Enter your height in centimeters: "))
elif unit1 == "inches":
    height = float(input("Enter your height in inches: "))
else:
    unit1 = input("Enter a valid unit only. (cm or inches): ")
    height = float(input(f"Enter your height in {unit1}: "))

#convert height to cm if entered in inches
if unit1 == "inches":
    height = height * 2.54
else:
    height = height

#ask for unit of weight and then value for weight
unit2 = input("Which unit would you like to enter your weight in? (kg or lbs): ")
if unit2 == "kg":
    weight = float(input("Enter your weight in kilograms: "))
elif unit2 == "lbs":
    weight = float(input("Enter your weight in pounds: "))
else:
    unit2 = input("Enter a valid unit only. (kg or lbs): ")
    weight = float(input(f"Enter your weight in {unit2}: "))

#convert weight to kg if entered in lbs
if unit2 == "lbs":
    weight = weight * 0.453592
else:
    weight = weight

#Calculate BMI
height_meters = height / 100
bmi = weight / (height_meters ** 2)

#Print BMI and weight status
if bmi < 18.5:
    print(f"Your BMI is {bmi:.2f}. You are underweight.")
elif bmi >= 18.5 and bmi <= 24.9:
    print(f"Your BMI is {bmi:.2f}. You are healthy weight.")
elif bmi >= 25 and bmi <= 29.9:
    print(f"Your BMI is {bmi:.2f}. You are overweight.")
elif bmi >= 30:
    print(f"Your BMI is {bmi:.2f}. You are obese.")
else:
    print("Please enter the valid data only!")
```
