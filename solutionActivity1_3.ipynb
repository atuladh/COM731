Utilise the dataset "2020_US_Election.csv" for data retrieval to gain insights into the specific requirements aimed at understanding the dynamics of the US election.

Import pandas

```python
# write your code here
import pandas as pd
```

Read dataset from csv file

```python
# Read dataset from csv file 
# If you save the Jupyter Notebook in the same location as your dataset, 
# you can simply use the filename when loading the data—there’s no need to specify the file path.

file_path = "2020_US_election.csv"
try:
    election_df=pd.read_csv(file_path, header=0, encoding='utf-8')
except FileNotFoundError as e1:
    print(f"File not found - {e1}")
except Exception as e2:
    print(f"Exception - {e2}")
```

```python
election_df
```

Slice every rows from the column 'candidate' to the end of the dataframe and assign the result to right_columns variable using `.loc[]`

```python
#Write your code here
# Slice every rows and the columns from the starting column to 'candidate': the end of the dataframe
right_columns = election_df.loc[:, 'candidate']

right_columns
```

Print the right_columns using pretty print to_markdown

```python
#Write your code here
# Print the output of right_columns.head()
print(f"Right Columns: \n {right_columns.to_markdown(index=False)}")
```

Create a new DataFrame to show the states with the most electors: California using `.loc[]`, and assign it to most_electors_state. Hint: You need to use an index property and column condition

```python
#Write your code here
index_most_electors = election_df[election_df['state']== 'CALIFORNIA'].index.tolist()

index_most_electors
```

```python
most_electors_state = election_df.iloc[index_most_electors]
most_electors_state
```

Print the most_electors_state DataFrame using pretty print to_markdown

```python
#Write your code here
print(most_electors_state.to_markdown(index=False))
```

Save the data from the 'most_electors_state' dataframe to a CSV file called California_votes.csv in a folder Data

```python
#Write your code here
most_electors_state.to_csv('Data/California_votes.csv')
```
