```python
#To check the basic salary of an employee based on their levels
salary = 0
level = input("Enter your level" )
if level == "Entry-level":
    salary = 30000
elif level == "Mid-level":
    salary = 50000
elif level == "Senior-level":
    salary = 70000
else: 
    salary = 0
    
if salary == 0:
    print("Invalid level entered.")
else:
    print(f"Your {level} salary is: {salary:}")
```

```python
#To check the basic salary of an employee based on their levels
salary = {"Entry-level": 30000, "Mid-level": 50000, "Senior-level": 70000}
level = input("Enter your level" )
salary = salary.get(level, 0)
    
if salary == 0:
    print("Invalid level entered.")
else:
    print(f"Your {level} salary is: {salary:}")
```
